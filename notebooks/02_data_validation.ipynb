```python
file_path = "../data/raw/health_raw_1960_2024_extended.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
required_columns = [
    "Country",
    "Country_Code",
    "Year",
    "Life_Expectancy"
]

for col in required_columns:
    if col in df.columns:
        print(f"✓ {col}")
    else:
        print(f"✗ Missing: {col}")
```

```python
print(df.dtypes)
```

```python
duplicates = df.duplicated(
    subset=["Country", "Year"]
).sum()

print("Duplicate Country-Year records:", duplicates)
```

```python
print("Minimum Year:", df["Year"].min())
print("Maximum Year:", df["Year"].max())

print("\nInvalid years:")
print(df[(df["Year"] < 1960) | (df["Year"] > 2024)])
```

```python
print("Number of countries:", df["Country"].nunique())

print("\nSample countries:")
print(df["Country"].unique()[:20])
```

```python
print("Minimum Life Expectancy:", df["Life_Expectancy"].min())
print("Maximum Life Expectancy:", df["Life_Expectancy"].max())

invalid_life = df[
    (df["Life_Expectancy"] < 0) |
    (df["Life_Expectancy"] > 120)
]

print("Invalid Life Expectancy records:", len(invalid_life))
```

```python
numeric_columns = df.select_dtypes(
    include=np.number
).columns

negative_values = {}

for col in numeric_columns:
    count = (df[col] < 0).sum()
    if count > 0:
        negative_values[col] = count

negative_values
```

```python
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_report = missing_report.sort_values(
    "Missing_Percentage",
    ascending=False
)

missing_report
```

```python
validation_summary = {
    "Total Rows": len(df),
    "Total Columns": len(df.columns),
    "Countries": df["Country"].nunique(),
    "Start Year": df["Year"].min(),
    "End Year": df["Year"].max(),
    "Duplicate Rows": df.duplicated().sum(),
    "Duplicate Country-Year": df.duplicated(
        subset=["Country", "Year"]
    ).sum(),
    "Missing Life Expectancy": df["Life_Expectancy"].isnull().sum()
}

validation_summary
```

```python
import os



output_path = "../data/processed/life_expectancy_validated.csv"

df.to_csv(output_path, index=False)

print("Validated dataset saved successfully!")
print(output_path)
```
