```python
import pandas as pd
import numpy as np
```

Load cleaned dataset

```python
file_path = "../data/processed/life_expectancy_clean.csv"

df = pd.read_csv(file_path)

print("Dataset loaded!")
print("Shape:", df.shape)
```

Check columns

```python
print(df.columns.tolist())
```

Create Age-Group Population Features

```python
df["Young_Population_Ratio"] = (
    df["Population_Young"] / df["Population"]
)

df["WorkingAge_Population_Ratio"] = (
    df["Population_WorkingAge"] / df["Population"]
)

df["Elderly_Population_Ratio"] = (
    df["Population_Elderly"] / df["Population"]
)
```
