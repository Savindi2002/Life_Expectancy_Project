```python
import pandas as pd
import numpy as np
```

Load cleaned dataset

```python
file_path = "../data/processed/life_expectancy_clean.csv"

df = pd.read_csv(file_path)

print("Dataset loaded!")
print("Shape:", df.shape)
```

Check columns

```python
print(df.columns.tolist())
```

Create Age-Group Population Features

```python
df["Young_Population_Ratio"] = (
    df["Population_Young"] / df["Population"]
)

df["WorkingAge_Population_Ratio"] = (
    df["Population_WorkingAge"] / df["Population"]
)

df["Elderly_Population_Ratio"] = (
    df["Population_Elderly"] / df["Population"]
)
```

Create Health Indicators

```python
df["Child_Mortality_Burden"] = (
    df["Infant_Mortality"] + df["Under5_Mortality"]
)

df["Health_Service_Access"] = (
    df["Basic_Water_Access"] +
    df["Basic_Sanitation_Access"] +
    df["Electricity_Access"]
) / 3
```

Create Economic Features

```python
df["GDP_Per_Capita_Growth"] = (
    df.groupby("Country")["GDP_Per_Capita"]
    .pct_change() * 100
)

df["Economic_Health_Index"] = (
    df["GDP_Per_Capita"] +
    df["GNI_Per_Capita"]
) / 2
```

Create Education Index

```python
education_columns = [
    "Primary_Enrollment",
    "Secondary_Enrollment",
    "Tertiary_Enrollment"
]

df["Education_Index"] = (
    df[education_columns].mean(axis=1)
)
```

Create Health Expenditure Index

```python
health_exp_columns = [
    "Health_Expenditure_GDP",
    "Government_Health_Expenditure",
    "OutOfPocket_Health_Expenditure"
]

df["Health_Expenditure_Index"] = (
    df[health_exp_columns].mean(axis=1)
)
```

Create Infrastructure Index

```python
infrastructure_columns = [
    "Electricity_Access",
    "Clean_Cooking_Access",
    "Internet_Users"
]

df["Infrastructure_Index"] = (
    df[infrastructure_columns].mean(axis=1)
)
```

Create Healthcare Resource Index

```python
healthcare_columns = [
    "Hospital_Beds",
    "Physicians"
]

df["Healthcare_Resource_Index"] = (
    df[healthcare_columns].mean(axis=1)
)
```

Create Year Feature

```python
df["Years_Since_1960"] = df["Year"] - 1960
```

Create decade

```python
df["Decade"] = (df["Year"] // 10) * 10

print(df[["Year", "Decade"]].head())
```

Check new features
